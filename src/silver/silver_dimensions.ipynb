### DimTrack

```python
df_track = spark.readStream.format("cloudFiles")\
                .option("cloudFiles.format","parquet")\
                .option("cloudFiles.schemaLocation" , "abfss://silver@nikunjazureproject.dfs.core.windows.net/DimTrack/checkpoint")\
                .option("schemaEvolutionMode","rescue")\
                .load("abfss://bronze@nikunjazureproject.dfs.core.windows.net/DimTrack")
```

```python
df2 = spark.read.format("parquet")\
           .load("abfss://bronze@nikunjazureproject.dfs.core.windows.net/DimTrack")
```

```python
display(df2)
```

```python
df_track = df_track.withColumn(
    "durationFlag",
    when(col('duration_sec') < 150, "low")
    .when(col('duration_sec') < 300, "medium")
    .otherwise("high")
)

df_track = df_track.withColumn("track_name", regexp_replace(col('track_name'), '-', ' '))
```

```python
df_track_obj = reusable()
df_track = df_track_obj.dropColumns(df_track,['_rescued_data'])
```

```python
df_track.writeStream.format("delta")\
    .outputMode("append")\
    .option("checkpointLocation","abfss://silver@nikunjazureproject.dfs.core.windows.net/DimTrack/checkpoint")\
    .trigger(once=True)\
    .option("path","abfss://silver@nikunjazureproject.dfs.core.windows.net/DimTrack/data")\
    .toTable("spotify_cata.silver.DimTrack")
```
